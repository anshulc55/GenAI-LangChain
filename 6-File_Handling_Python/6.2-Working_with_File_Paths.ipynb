## <span style="color:yellow"> Working with File Paths in Python </span>
When working with files in Python, handling file paths correctly is crucial to ensure your code works across different operating systems and environments. 

Python provides several modules and functions for working with file paths effectively.

### <span style="color:lightgreen"> Modules for File Path Management </span>
Python has built-in modules that handle file paths and directory operations.
1. <span style="color:green"> __os.path__ </span> : Basic module for file path operations.

2. <span style="color:green">__pathlib__</span>: More modern and recommended, introduced in Python 3.4.

### <span style="color:lightgreen"> Using os.path for File Paths </span>

```python
# Getting the Current Working Directory:
import os
print(os.getcwd())
```

```python
# Joining Paths
# Use os.path.join to join multiple path segments.

import os
path = os.path.join("/Users", "anshul","Training", "GenAI-LangChain","6-File_Handling_Python")
print(path)

# Checking Path Existence
print(os.path.exists(path + "/SampleDir"))

# File or Directory Check
print(os.path.isfile(path+"/SampleFile2.txt"))
print(os.path.isdir(path))
```

### <span style="color:lightgreen"> Using pathlib for File Paths </span>

```python
from pathlib import Path

# Getting the Current Directory:
print(Path.cwd())

# Joining Paths
path = Path("/Users") / "anshul" / "Training" / "GenAI-LangChain" / "6-File_Handling_Python"
print(path)

print(path.exists())
print(path.is_file())
print(path.is_dir())
```

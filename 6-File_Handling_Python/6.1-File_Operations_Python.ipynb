## <span style="color:yellow"> File Operations with Python </span>
File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. 

This lesson will cover the basics of file handling, including reading and writing text files and binary files.

### <span style="color:lightgreen"> File Modes </span>
Python offers different modes to interact with files:
1. <span style="color:green"> __'r': Read (default)__ </span> : Opens a file for reading. Error if the file does not exist.

2. <span style="color:green">__'w': Write__</span>: Creates a new file for writing, or truncates the file if it exists.

3. <span style="color:green">__'a': Append__</span>: Opens the file for appending. Creates the file if it doesn’t exist.

4. <span style="color:green">__'r+': Read and Write__</span>: Opens the file for both reading and writing.

5. <span style="color:green">__'b': Binary mode__</span>: used with any of the above modes for binary files.

### <span style="color:lightgreen"> Basic File Operations </span>

```python
# Opening a File

file = open("SampleFile.txt", "r")
print("Successfully able to Read the File...")
file.close()
```

```python
# Open file Using the with Statement
# The with statement automatically closes the file after the block is executed, even if an error occurs.

with open("SampleFile.txt", "r") as file:
    print("Successfully Opened the file...")
```

### <span style="color:lightgreen"> Reading from Files </span>

```python
# read(): Reads the entire content.
with open("SampleFile.txt", "r") as file:
    fileContent = file.read()
    print(fileContent)
```

```python
# readline(): Reads one line at a time.

with open("SampleFile.txt", "r") as file:
    fileContent = file.readline()
    print(fileContent)
```

```python
# readlines(): Reads all lines and returns them as a list.
with open("SampleFile.txt", "r") as file:
    fileContent = file.readlines()
    print(fileContent)
```

```python
# Read file and Print all Lines
with open("SampleFile.txt", "r") as file:
    for line in file:
        print(line.strip())
```

### <span style="color:lightgreen"> Writing to Files </span>

```python
# write(): Writes a single string.
with open("SampleFile1.txt", "a") as file:
    file.write("Hello, World!\n")
    file.write("Writing to a file in Python.")
```

```python
# writelines(): Writes a list of strings.

# List of lines to be written to the file
lines = [
    "This is the first line.\n",
    "This is the second line.\n",
    "This is the third line.\n"
]

with open("SampleFile2.txt", "w") as file:
    file.writelines(lines)
    
```

### <span style="color:lightgreen"> Working with Binary Files </span>

```python
# Binary files handle non-text data like images.
with open("/Users/anshul/Downloads/selenium.jpeg", "rb") as file:
    content = file.read()
    print("Read binary content:", content[:20])
```

### <span style="color:lightgreen"> Exception Handling in File Operations </span>

```python
# Use try and except to handle exceptions during file operations.
try:
    with open("/Users/anshul/Downloads/selenium12.jpeg", "rb") as file:
        content = file.read()
        print("Read binary content:", content[:20])
except FileNotFoundError:
    print("File not found. Please check the file path.")
```

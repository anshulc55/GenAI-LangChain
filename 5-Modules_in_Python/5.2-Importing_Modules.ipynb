## Importing Modules in Python
In Python, importing modules allows us to use external modules, be it built-in or user-defined code, in our programs. 

Importing provides access to functions, classes, and variables defined in modules. Python’s import statement is flexible and provides multiple ways to bring modules into the current namespace.

### Key Ways to Import Modules:

#### Basic Import
With a basic import, we bring the entire module into our program and access its members using dot notation

```python
# import <Module_name>

import math, os

math.sqrt(25)
math.pi
math.sin(90)

os.cpu_count()
os.getcwd()
```

#### Selective Import
Selective import allows us to import specific functions, classes, or variables from a module. This is particularly useful when we only need certain parts of a module.

***from module_name import specific_name***

```python
# from module_name import specific_name

from math import sqrt, pi

print("Square root of 25:", sqrt(25))
print("Value of pi:", pi)
```

#### Aliased Import
Aliased import gives a module an alias (nickname) to simplify its usage, especially when dealing with long or frequently used module names. This technique improves readability, especially for commonly used modules like numpy and pandas.

___import module_name as alias_name___

```python
# import module_name as alias_name

import math as m

m.sqrt(25)
m.pi
```

#### Import All (Wildcard Import)
The wildcard import (*) imports all public names from a module into the current namespace. While this might seem convenient, it can lead to name conflicts and is generally discouraged, especially in larger codebases.

___from module_name import *___

```python
# from module_name import *

from math import *

print("Square root of 25:", sqrt(25))
print("Value of pi:", pi)
```

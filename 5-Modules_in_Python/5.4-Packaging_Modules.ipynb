## <span style="color:yellow"> Packaging Modules </span>
In Python, packaging modules is a way to organize related modules into a single distribution that can be easily shared and installed. 

A package typically contains one or more modules, along with additional metadata about the package. This makes it easy to distribute your code and reuse it across different projects.

### <span style="color:lightgreen"> Key Concepts of Packaging </span>
1. <span style="color:green"> __Packages__ </span> : A package is a directory that contains a special __init__.py file, which indicates that the directory should be treated as a package. This file can be empty or can contain initialization code for the package.

2. <span style="color:green">__Module__</span>: A module is a single Python file (.py) that contains functions, classes, or variables.

3. <span style="color:green">__Distribution__</span>: A package distribution can be created to share your package with others, often via the Python Package Index (PyPI).

### <span style="color:lightgreen"> Structure of a Package </span>

A typical package structure might look like this:

```python
math_operations/
│
├── addition.py
├── subtraction.py
├── __init__.py
└── setup.py
```

### <span style="color:lightgreen"> Using the Package </span>

```python
import mathlib

# Using arithmetic functions
print("Addition: ", mathlib.add(20, 40))
print("Subtraction:", mathlib.subtract(15, 3))
print("Multiplication:", mathlib.multiply(25, 30))
print("Division:", mathlib.divide(6, 3))

# Using Geometry functions
print("Area of Circle (radius 5):", mathlib.area_circle(5))
print("Perimeter of Circle (radius 5):", mathlib.perimeter_circle(5))
print("Area of Rectangle (length 4, width 6):", mathlib.area_rectangle(4, 6))
print("Perimeter of Rectangle (length 4, width 6):", mathlib.perimeter_rectangle(4, 6))
```

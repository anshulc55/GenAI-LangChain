## <span style="color:yellow"> Using Third-Party Modules in Python </span>
Third-party modules in Python are libraries developed by the Python community and shared publicly. 

These modules extend the functionality of Python and can be installed via pip, the Python package manager. 

Third-party modules cover a vast range of needs—from web development and data manipulation to machine learning and automation—making Python extremely versatile.

### <span style="color:lightgreen"> Why Use Third-Party Modules? </span>
1. <span style="color:green"> __Extend Python’s Core Capabilities__ </span> : Add specialized functionalities that aren’t available in the standard library.

2. <span style="color:green">__Save Development Time__</span>: Use pre-built, tested libraries instead of writing new code from scratch.

3. <span style="color:green">__Collaborate with the Community__</span>: Benefit from open-source contributions and updates, making your code more robust and current.

### <span style="color:lightgreen"> Installing Third-Party Modules </span>

You can install third-party modules using pip. The general syntax is:

<span style="color:red"> pip install <module_name> </span>

### <span style="color:lightgreen"> Commonly Used Third-Party Modules with Examples </span>

```python
# requests – HTTP Library for API Requests

import requests

# Making a GET request to a public API
#response = requests.get("https://api.github.com")
response = requests.get("https://www.facebook.com")

# Checking if the request was successful
if response.status_code == 200:
    # Accessing the JSON content
    #data = response.json()
    data = response
    print("API Data:", data.content)
else:
    print("Failed to retrieve data", response.status_code)
```

```python
# pandas – Data Manipulation and Analysis

import pandas

# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [24, 27, 22],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}

# Displaying the DataFrame
data_frame = pandas.DataFrame(data)
print("DataFrame ::")
print(data_frame)

# Adding a new column
data_frame['Score'] = [85, 90, 88]
print("\nUpdated DataFrame:")
print(data_frame)
```

```python
# numpy – Numerical Computation

import numpy as np

# Creating an array
array = np.array([1, 2, 3, 4, 5])

# Array operations
print("Original array:", array)
print("Array multiplied by 2:", array * 2)
print("Square root of each element:", np.sqrt(array))

# Creating a 2D array (matrix)
matrix = np.array([[1, 2], [3, 4]])
print(matrix)

# Matrix multiplication
result = np.dot(matrix, matrix)
print(result)

```

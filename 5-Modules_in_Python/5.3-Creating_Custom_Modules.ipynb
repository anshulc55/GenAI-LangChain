## Creating Custom Modules in Python
In Python, creating a custom module is simple: it’s just a .py file that contains code you want to reuse across different scripts.

#### Creating a Simple Custom Module

```python
import customMath, math

print("Sum:", customMath.add(10, 30))
print("Difference:", customMath.subtract(100, 30))

# Accessing a variable from the module
print("Value of PI:", customMath.PI)

print("Value of PI from MATH:", math.pi)
```

#### Creating a Custom Module with Classes

```python
from customShapes import Circle, Rectangle

circle = Circle(6)
print("Area of Circle is :", circle.area())
print("Circumference of the circle:", circle.circumference())

rectangle = Rectangle(10, 20)
print("Area of Rectangle is :", rectangle.area())
```

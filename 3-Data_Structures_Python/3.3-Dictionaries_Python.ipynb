## Dictionaries in Python

Dictionaries are powerful data structures used to store data in key-value pairs 

__Key Features of Dictionaries:__

*Unordered*: Dictionaries do not maintain the order of elements.

*Mutable*: You can modify, add, or remove key-value pairs after creating a dictionary.

*Unique Keys*: Each key must be unique and immutable (e.g., strings, numbers, tuples).


### Creating Dictionaries
You can create dictionaries using curly braces {}, with each key-value pair separated by a colon :.

```python
# Python Dictionary

my_dict = {"Key1": "Value1", "Key2" : "Value2"}

person = {"name": "John", "age": 30, "city": "New York"}
print(person)

# Empty Dictionary
empty_Dict = {}
print(type(empty_Dict))

# Dictionary with mixed types
mixed_dict = {"name": "Alice", "age": 25, 1: [2, 4, 3]}
print(mixed_dict)
```

### Accessing Dictionary Elements
You can access the value of a dictionary by using its key. Use square brackets [] or the get() method for safe access without errors.

```python
person = {"name": "John", "age": 30, "city": "New York"}

# Accessing values

print(person["name"])
print(person.get("age"))

print(person.get("agee", "Key Missing, Defaul is GREEN"))

# Modify values
person["city"] = "Tokyo"
print(person)
```

### Adding and Removing Key-Value Pairs

You can add new key-value pairs or remove existing pairs with ease.

```python
# Removing Key-Value Pairs
person = {"name": "John", "age": 30, "city": "New York"}

# Add the values in Dictionary 
person["phone_number"] = "+1 - 9876568765"
print(person)

# Removing a key-value pair
del person["age"]
print(person)

# Using pop() to remove and return the value
print(person.pop("phone_number"))
print("**************")
print(person)
```

### Dictionary Methods
Python provides several built-in methods for dictionaries.

__Common Dictionary Methods:__

clear(): Removes all items.

copy(): Returns a shallow copy.

keys(): Returns a view object of all keys.

values(): Returns a view object of all values.

items(): Returns a view object of all key-value pairs.

pop(): Removes the specified key and returns the value.

popitem(): Removes and returns the last inserted key-value pair (from Python 3.7+).

update(): Updates the dictionary with another dictionary or key-value pairs.

```python
person = {"name": "John", "age": 30, "city": "New York"}

print(person.keys())
print(person.values())
print(person.items())

for item in person.values():
    print(item)

person.update({"country": "USA", "Citizen": "Yes"})
print(person)

person.clear()
print(person)
```

### Iterating Over Dictionaries
You can loop through dictionaries using keys(), values(), and items() to iterate over keys, values, or both.



```python
person = {"name": "John", "age": 30, "city": "New York"}

# Iterating over keys
for key in person.keys():
    print(person.get(key))
print("******************")

# Iterating over Values
for key in person.values():
    print(key)
print("******************")


# Iterating over Items
for key in person.items():
    print(key)
print("******************")
```

### Nested Dictionaries
Dictionaries can store other dictionaries, allowing for the creation of complex data structures like hierarchical data.

```python
company = {
    "employee_1": {"name": "Alice", "age": 28, "role": "Engineer"},
    "employee_2": {"name": "Bob", "age": 32, "role": "Manager"}
}

print(company["employee_2"])
print(company["employee_2"]["role"])
```

## Tuples in Python

Tuples are another built-in data structure in Python, similar to lists, but with one major difference: tuples are immutable. 
Once a tuple is created, its contents cannot be changed. 

__Key Features of Tuples:__

*Ordered*: Elements in a tuple maintain the order in which they are added.

*Immutable*: Once a tuple is created, its elements cannot be changed, added, or removed.

*Heterogeneous*: A tuple can contain elements of different types, just like a list.


### Creating Tuples
Tuples are created by enclosing elements in parentheses () and separating them with commas.

```python
# list -- []
# tuple -- ()

my_tuple = (1, 100 , "James", 30.4  , True, (3, 4, "Maria"), ["List", "Python", 11])
print(type(my_tuple))
print(my_tuple)
```

### Accessing Tuples Elements
You can access elements in a tuple using indexing. Python uses zero-based indexing, so the first element has an index of 0.

```python
my_tuple = (1, 100 , "James", 30.4  , True, (3, 4, "Maria"), ["List", "Python", 11])

print(my_tuple[5])
print(my_tuple[3])

print(my_tuple[-3])
print(my_tuple[3:5])
```

### Modifying Tuple
Once created, a tuple cannot be modified. This includes adding, removing, or changing elements.

```python
my_tuple = (1, 100 , "James", 30.4  , True, (3, 4, "Maria"), ["List", "Python", 11])
my_list = [1, 100 , "James", 30.4  , True, (3, 4, "Maria"), ["List", "Python", 11]]

my_list[3] = "Updated Value"
print(my_list)

my_tuple[3] = "Updated Value"
print(my_tuple)
```

### Tuple Methods
Tuples are immutable, they have fewer methods compared to lists.

__Common List Methods:__

count(): Returns the number of occurrences of a specified value in the tuple.

index(): Returns the index of the first occurrence of a specified value.

```python
numbers = (1, 2, 3, 2, 2, 5, 2, 2, 22,4)

print(numbers.count(2))
print(numbers.index(4))
```

### Advantages of Tuples Over Lists
Immutability: This ensures that data cannot be accidentally modified, making tuples a great choice when you need to protect data integrity.

Performance: Tuples are faster than lists when it comes to iteration and access because of their immutability.

Memory Efficiency: Tuples use less memory compared to lists, making them more efficient in memory-sensitive applications.




### Conversion Between Lists and Tuples
You can convert a list to a tuple using the tuple() function, and a tuple to a list using the list() function.

```python
my_list = [1, 2, 3, 4]
my_tuple = (1, 2, 3, 4)

print (f"Type of my_list before conversion {type(my_list)}")
print (f"Type of my_tupe before conversion {type(my_tuple)}")

print("*******************")
my_list = tuple(my_list) # List to Tuple Conversion
my_tuple = list(my_tuple) # Tuple to List Conversion

print (f"Type of my_list After conversion {type(my_list)}")
print (f"Type of my_tupe After conversion {type(my_tuple)}")
```

### Common Mistakes and Errors
Modifying Tuple Elements: Trying to modify an element in a tuple will raise an error because tuples are immutable.

Confusing Single-Element Tuples: Forgetting the comma when creating a single-element tuple can result in unexpected behavior.


```python
my_tuple = (10,)
print(type(my_tuple))
```

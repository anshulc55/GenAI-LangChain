## <span style="color:yellow"> Building Interactive Web Apps for Data Science and Machine Learning </span>
We will learn about the Machine Learning App Deployment Model "Streamlit".

Using **streamlit** you can deploy any machine learning model and any python project with ease and without worrying about the frontend. Streamlit is very user-friendly.

We will learn some important functions of streamlit, create a python project, and deploy the project on a local web server.

##### <span style="color:skyblue"> Install StreamLit </span>
**pip install streamlit**


### <span style="color:lightgreen"> How to Run Streamlit App </span>
**streamlit run filename.py**

## <span style="color:yellow"> Use "self" in Python class </span>
"self" is a reference to the current instance of the class. It allows access to the instance’s attributes and other methods.

Using self, each instance of a class can maintain its own set of attributes, enabling instance-specific data storage.

Without self, it would be difficult for methods within a class to know which specific instance's attributes they should access or modify. self ensures that each object keeps its state separate from others.

### <span style="color:orange"> How self is Used in Class Methods </span>
In Python, every instance method in a class must have self as its first parameter. This is how Python can differentiate between the attributes of different instances.

```python
class Car:
    # Constructor that initializes the instance attributes using self
    def __init__(self, make, model):
        self.make = make    # self.make is an instance attribute
        self.model = model

    # Method to display car information
    def display_info(self):
        print(f"Car Make: {self.make}, Model: {self.model}")

# Creating two instances of the Car class
car1 = Car("Toyota", "Camry")
car2 = Car("Honda", "Accord")

car1.display_info()
car2.display_info()
```

### <span style="color:lightgreen"> self in Different Methods </span>

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def birthday(self):
        self.age += 1
        print(f"Happy Birthday {self.name}! You are now {self.age} years old.")

# Creating an instance of Person
person1 = Person("Alice", 30)
person2 = Person("Peter", 50)

print(f"Name : {person1.name}, Age : {person1.age}")
person1.birthday()
print(f"Name : {person1.name}, Age : {person1.age}")
print("*************************************")
print(f"Name : {person2.name}, Age : {person2.age}")
```

### <span style="color:lightgreen"> self to Call Other Methods </span>
Using self, you can call one method from another within the same class. This is particularly useful for modularizing code within the class.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    # Method to calculate area
    def area(self):
        return self.width * self.height
    
    # Method to calculate perimeter
    def perimeter(self):
        return 2 * (self.area())
    
    # Method to display both area and perimeter by calling other methods
    def display_info(self):
        # Calls area() using self
        print(f"Area: : {self.area()}")
        print(f"Perimeter: : {self.perimeter()}")

rectange1 = Rectangle(10,5)
rectange1.display_info()
```

### <span style="color:orange"> Example of How self Keeps Instances Separate </span>

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def give_raise(self, amount):
        self.salary += amount
        print(f"{self.name}'s new salary is {self.salary}")

# Creating two instances of Employee
emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)

emp1.give_raise(10000)
emp2.give_raise(5000)
```

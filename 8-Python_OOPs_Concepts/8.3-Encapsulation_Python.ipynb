## <span style="color:yellow"> Encapsulation in Python </span>
Encapsulation is one of the fundamental concepts in object-oriented programming (OOP). It describes the idea of wrapping data and the methods that work on data within one unit. 

This puts restrictions on accessing variables and methods directly and can prevent the accidental modification of data.

A class is an example of encapsulation as it encapsulates all the data that is member functions, variables, etc. 

<span style="color:skyblue"> **Purpose** </span>
1. Protect the data from unauthorized access.
2. Improve code modularity and maintainability.

### <span style="color:orange"> How Encapsulation Works in Python </span>
Encapsulation is implemented using classes and objects.

Data (variables) and methods are encapsulated together in a class.

### <span style="color:lightgreen"> Access Modifiers in Python </span>
Python doesn't have explicit keywords like private, protected, or public. Instead, it uses naming conventions:

<span style="color:yellow"> **Public** </span> : Accessible from anywhere (default).

<span style="color:yellow"> **Protected** </span> : Indicated by a single underscore (_); meant to be accessed only within the class or subclass.

<span style="color:yellow"> **Private** </span> : Indicated by a double underscore (__); name mangling restricts access to within the class.

```python
# Public Access Modifier
class Car:

    def __init__(self, brand):
        self.brand = brand  # Public attribute

car = Car("Toyota")
print(car.brand)
```

```python
# Protected Access Modifier

class Animal:

    def __init__(self, species):
        self._species = species  # Protected attribute

class Dog(Animal):
    def __init__(self, species, breed):
        self.breed = breed
        super().__init__(species)

dog = Dog("Mammal", "Labrador")
print(dog.breed)
# print(dog._species)   # Accessible but not recommended
```

```python
# Private Access Modifier

class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number   # Private attribute
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount("123456", 1000)
print(account.get_balance())
# print(account.__balance)
```

### <span style="color:orange"> Private Methods in Python </span>
Private methods are methods within a class that are not intended to be accessed directly from outside the class.

<span style="color:skyblue"> **Purpose** </span>
Restrict access to methods that should only be used internally by the class.

Enhance encapsulation and prevent unintended usage.

```python
# ATM System

class ATM:

    def __init__(self, balance):
        self.__balance = balance
    
    def __authenticate_user(self, pin):
        # Authentication logic here
        print("Authentication Success!!!")
        return True

    def withdraw(self, amount, pin):
        if self.__authenticate_user(pin):
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Withdrawn: {amount}. Remaining balance: {self.__balance}")
            else:
                print("Insufficient balance")

atm = ATM(10000)
#atm.__authenticate_user(1234)
atm.withdraw(4000, 1234)

```

#### <span style="color:lightgreen"> How Private Methods Work </span>
**Name Mangling**: Private methods are internally renamed by Python to _ClassName__MethodName.

This mechanism prevents accidental method overrides and restricts external access.

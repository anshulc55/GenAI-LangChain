## <span style="color:yellow"> Polymorphism in Python </span>
Polymorphism means "many forms." In programming, it allows objects of different classes to be treated as objects of a common superclass. In Python, polymorphism can be achieved in several ways, including:

<span style="color:skyblue"> Duck Typing </span>

<span style="color:skyblue"> Operator Overloading </span>

<span style="color:skyblue"> Method Overriding </span>

<span style="color:skyblue"> Method Overloading </span>


### <span style="color:orange"> Duck Typing in Python </span>
In Python, polymorphism is based on the principle:
"If it walks like a duck and quacks like a duck, it must be a duck."

This means that Python doesn't care about the actual type of an object; it cares only about its behavior.

```python
class Dog:
    def speak(self):
        return "Woof!"
    
class Cat:
    def speak(self):
        return "Meow!"

class Bird:
    def speak(self):
        return "Chirp!"

def animal_sound(animal):
    print(animal.speak())

# Using Duck Typing
dog = Dog()
cat = Cat()
bird = Bird()

animal_sound(dog)
animal_sound(cat)
animal_sound(bird)
```

### <span style="color:lightgreen"> Operator Overloading </span>
Operator overloading in Python allows us to define custom behavior for standard operators (+, -, *, /, etc.) when they are applied to objects of a user-defined class. 

By overriding special methods (also called "magic methods"), we can customize how operators work with instances of our class.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"

# Using Operator Overloading
point1 = Point(1, 2)
point2 = Point(3, 4)
result = point1 + point2
print(result)  # Point(4, 6)
```

### <span style="color:orange"> Method Overriding </span>
Method overriding is a feature in object-oriented programming where a subclass provides a specific implementation of a method that is already defined in its superclass. 

This allows a subclass to modify or extend the behavior of methods inherited from a parent class without changing the parent class itself.


##### <span style="color:skyblue"> Basic Method Overriding </span>


```python
class Vehicle:
    def move(self):
        return "The vehicle is moving."

class Car(Vehicle):
    def move(self):
        return "The car is driving on the road."

class Bike(Vehicle):
    def move(self):
        return "The bike is pedaling."

# Create instances of each class
vehicle = Vehicle()
car = Car()
bike = Bike()

# Call the move method
print(vehicle.move())  # The vehicle is moving.
print(car.move())      # The car is driving on the road.
print(bike.move())     # The bike is pedaling.
```

##### <span style="color:skyblue"> Method Overriding with Arguments </span>


```python
class Animal:
    def speak(self):
        return "Some generic animal sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

class Cow(Animal):
    def speak(self, times):
        return "Moo! " * times

# Create instances
animals = [Animal(), Dog(), Cat(), Cow()]

# Print the speak output for each animal
for animal in animals:
    if isinstance(animal, Cow):  # Cow requires an argument for 'times'
        print(animal.speak(3))  # Moo! Moo! Moo!
    else:
        print(animal.speak())  # Some generic animal sound, Woof!, Meow!
```

##### <span style="color:skyblue"> Method Overriding with super() </span>


```python
class Animal:
    def speak(self):
        return "Some generic animal sound"

class Dog(Animal):
    def speak(self):
        parent_speech = super().speak()  # Call the parent class method
        return f"{parent_speech} but specifically a dog says Woof!"

# Create an instance of Dog
dog = Dog()
print(dog.speak())  # Some generic animal sound but specifically a dog says Woof!
```

##### <span style="color:skyblue"> Overriding Constructor (__init__) Method </span>


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound."

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Initialize the parent class (Animal)
        self.breed = breed

    def speak(self):
        return f"{self.name} the {self.breed} says Woof!"

# Create a Dog instance
dog = Dog("Buddy", "Golden Retriever")
print(dog.speak())  # Buddy the Golden Retriever says Woof!
```

## <span style="color:yellow"> Inheritance in Python </span>
Inheritance is a key feature of Object-Oriented Programming (OOP) that allows one class (child class) to inherit properties and behaviors (attributes and methods) from another class (parent class). It promotes code reuse and modularity.

<span style="color:skyblue"> **Purpose** </span>
1. Code Reusability: Avoid redundancy by using common functionality defined in the parent class.
2. Modularity: Organize code into hierarchical relationships.
3. Extensibility: Easily extend functionalities for specialized needs.

### <span style="color:orange"> Syntax of Inheritance </span>

```python
class Parent:
    # Parent class implementation

class Child(Parent):
    # Child class implementation
```

### <span style="color:lightgreen"> Basic Example of Inheritance </span>

```python
# A company has general employees and specialized roles like Developer and Manager.
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def get_details(self):
        return f"Name: {self.name}, Salary: ${self.salary}"

class Developer(Employee):
    def __init__(self, name, salary, programming_languages):
        self.programming_languages = programming_languages
        super().__init__(name, salary)

    def get_details(self):
        base_details = super().get_details()
        return f"{base_details}, Programming Languages: {', '.join(self.programming_languages)}"
    
class Manager(Employee):
    def __init__(self, name, salary, team_size):
        self.team_size = team_size
        super().__init__(name, salary)
    
    def get_details(self):
        base_details = super().get_details()
        return f"{base_details}, Team Size: {self.team_size}"
    
dev = Developer("Alice", 80000, ["Python", "JavaScript"])
print(dev.get_details())

mgr = Manager("Bob", 90000, 10)
print(mgr.get_details())  
```

#### <span style="color:lightgreen"> Python super() </span>
super() function is used to refer to the parent class or superclass. 

It allows you to call methods defined in the superclass from the subclass, enabling you to extend and customize the functionality inherited from the parent class.

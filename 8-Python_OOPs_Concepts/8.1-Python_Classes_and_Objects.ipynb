## <span style="color:yellow"> Object-Oriented Programming in Python </span>
Python object-oriented Programming (OOPs) is a programming paradigm that uses objects and classes in programming. 

It aims to implement real-world entities like inheritance, polymorphisms, encapsulation, etc. in the programming. 

### <span style="color:orange"> Python Classes and Objects </span>
classes and objects form the fundamental building blocks of object-oriented programming (OOP).

<span style="color:Green"> __Class__ </span> : A blueprint or template for creating objects. It defines the properties (attributes) and behaviors (methods) that objects created from the class will have.

<span style="color:Green"> __Object__ </span> : An instance of a class. Each object has its own set of attributes and can perform actions defined by its class methods.

```python
# Defining a class

class Dog:
    # Constructor method to initialize an object
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def display_info(self):
        print(f"My name is {self.name} and I am a {self.breed}.")


dog1 = Dog("Buddy", "Golden Retriever")
print(dog1.breed)
print(dog1.name)
print(dog1.display_info())
```

### <span style="color:lightgreen"> Attributes and Methods in a Class </span>
<span style="color:skyblue"> __Attributes__ </span> : Characteristics of the object, like name and breed in the Dog class. They define the data stored in an object.

<span style="color:skyblue"> __Methods__ </span> : Functions defined within a class that perform actions. Methods operate on the object’s attributes.

```python
class Dog:
    # Constructor method to initialize an object
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        print("Hi, I am awake")

    def display_info(self):
        print(f"My name is {self.name} and I am a {self.breed}.")
    
    def bark(self):
        print(f"{self.name} says Woof!")

dog1 = Dog("Buddy", "Golden Retriever")
dog1.bark()
```

### <span style="color:lightgreen"> __ init __ Method (Constructor) </span>
The "__ init __" method is a special method called a constructor. It is automatically invoked when a new object is created, allowing for the initialization of object attributes.

### <span style="color:lightgreen"> self Parameter </span>
Every instance method (including __ init __) takes self as the first parameter. self refers to the current instance of the class, allowing access to the attributes and methods of that instance.

### <span style="color:lightgreen"> Class vs Instance Attributes (Constructor) </span>
<span style="color:skyblue"> __Instance Attributes__ </span> : Defined within methods, typically initialized in the __ init __ method. They are specific to each instance of the class.

<span style="color:skyblue"> __Class Attributes__ </span> : Defined directly within the class body, shared across all instances of the class.

```python
class Circle:

    PI = 3.14444

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.PI * (self.radius**2)
    
# Creating instances of Circle
circle1 = Circle(10) 
circle2 = Circle(12)
circle3 = Circle(15)

print(f"Area of {circle1.radius} is : {circle1.area()}")
print(f"Area of {circle2.radius} is : {circle2.area()}")
print(f"Area of {circle3.radius} is : {circle3.area()}")

print(circle1.PI)
print(circle2.PI)
```

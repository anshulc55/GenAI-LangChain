## <span style="color:yellow"> Abstraction in Python </span>
Abstraction is a core concept in Object-Oriented Programming (OOP) that helps in hiding the complex implementation details from the user while exposing only the essential features. 

Abstraction is the process of simplifying complex systems by focusing on the essential features while hiding unnecessary details. It allows the programmer to concentrate on high-level logic while delegating complex operations to the underlying system.

##### <span style="color:skyblue"> Abstract classes </span>
A class that cannot be instantiated directly but serves as a blueprint for other classes.

A class that contains one or more abstract methods. It provides a common interface for subclasses.

##### <span style="color:skyblue"> Abstract methods </span>
Methods that are declared but contain no implementation. Subclasses of the abstract class must implement these methods.

A method that is declared but contains no implementation. The subclass must provide the implementation.


### <span style="color:lightgreen"> Abstract Classes and Methods in Python </span>
In Python, we use the abc (Abstract Base Class) module to define abstract classes. The ABC class and the abstractmethod decorator are used to create abstract classes and methods.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    
    @abstractmethod
    def area(self):
        pass
    
    @abstractmethod
    def perimeter(self):
        pass
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return self.area() * 2
    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius**2)
    
    def perimeter(self):
        return 2 * math.pi * self.radius
    

# Create instances of Rectangle and Circle
rect = Rectangle(4, 5)
circle = Circle(3)

print("Rectangle:")
print(f"Area: {rect.area()}")
print(f"Perimeter: {rect.perimeter()}")

print("\nCircle:")
print(f"Area: {circle.area()}")
print(f"Perimeter: {circle.perimeter()}")
```

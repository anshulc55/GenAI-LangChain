## <span style="color:yellow"> *args and **kwargs in Python </span>
In Python, functions can take a variable number of arguments. Two common ways to handle such cases are through the use of *args and **kwargs. These allow us to write functions that are flexible and can accept any number of positional and keyword arguments. 


### <span style="color:orange"> What is *args? </span>
The *args syntax in a function definition is used to pass a variable number of non-keyword arguments (positional arguments) to a function. 

The *args parameter collects any extra positional arguments passed to the function and stores them as a tuple.

**Flexibility**: Allows a function to accept any number of positional arguments.

**Convenience**: Avoids the need to specify every possible argument in the function signature.

```python
# Logging System

def log_message(level, *args):
    print(f"{level} : ", end="" )
    for arg in args:
        print(arg, end=" ")
    print()

log_message("INFO", "Server started successfully")
log_message("ERROR", "Connection failed", "Timeout", "Database")
log_message("DEBUG", "Database query executed", "Row count: 42")
```

### <span style="color:lightgreen"> What is **kwargs? </span>
The **kwargs syntax is used to pass a variable number of keyword arguments to a function. The **kwargs collects these keyword arguments into a dictionary, where the keys are the argument names, and the values are the argument values.

```python
def api_call(endpoint, **kwargs):
    print(f"Calling {endpoint} with the following parameters:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

api_call("getUser", user_id=101, fields="name,email")

api_call("getOrders", status="shipped", page=2)
```

### <span style="color:orange"> Combining *args and **kwargs </span>


```python
def display_info(name, *args, **kwargs):
     print(f"Name: {name}")
     print("Other Details:", args)
     print("Additional Info:", kwargs)

display_info("John", 25, "Engineer", location="New York", status="Active")
```

### <span style="color:orange"> Order of Parameters in Function Definition </span>
When defining a function that uses both *args and **kwargs, the order of parameters matters:

1. First, you should define any regular positional arguments.
2. Then, include *args for any additional positional arguments.
3. Finally, **kwargs should come last, as it accepts keyword arguments.

## <span style="color:yellow"> Multiple Inheritance in Python </span>
In multiple inheritance, a class can inherit attributes and methods from more than one parent class. This enables the derived class to have properties of all the parent classes.

### <span style="color:orange"> Syntax of Multiple Inheritance </span>

```python
class Parent1:
    # Attributes and methods of Parent1

class Parent2:
    # Attributes and methods of Parent2

class Child(Parent1, Parent2):
    # Child class inherits from Parent1 and Parent2
```

```python
# Multiple Inheritance

class Teacher:
    def teach(self):
        return "Teaching subjects."
    
class SportsCoach:
    def coach(self):
        return "Coaching sports."
    
class SchoolStaff(Teacher, SportsCoach):
     def manage(self):
        return "Managing school activities."

# Testing multiple inheritance
staff = SchoolStaff()
print(staff.manage())
print(staff.coach())
print(staff.teach())
```

### <span style="color:orange"> Diamond Problem in Multiple Inheritance </span>
The diamond problem arises when a class inherits from two classes that share a common ancestor. Python resolves this using Method Resolution Order (MRO).

```python
class A:
    def message(self):
        return "Message from A"

class B(A):
    def message(self):
        return "Message from B"

class C(A):
    def message(self):
        return "Message from C"

# D inherits from both B and C
class D(C, B):
    pass

# Testing diamond problem resolution
d = D()
print(d.message())  # Output: Message from B
print(D.mro())      # Displays the method resolution order
```

#### <span style="color:lightgreen"> Multilevel Inheritance </span>
In multilevel inheritance, a class inherits from a child class, which itself inherits from another parent class. This creates a chain of inheritance.


```python
class GrandParent:
    # Attributes and methods of GrandParent

class Parent(GrandParent):
    # Attributes and methods of Parent

class Child(Parent):
    # Attributes and methods of Child
```

```python
class Device:
    def __init__(self, brand):
        self.brand = brand

    def show_brand(self):
        return f"Device Brand: {self.brand}"


class Smartphone(Device):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def show_model(self):
        return f"Smartphone Model: {self.model}"
    

class Android(Smartphone):
    def __init__(self, brand, model, version):
        super().__init__(brand, model)
        self.version = version

    def show_version(self):
        return f"Android Version: {self.version}"
    
android_phone = Android("Samsung", "Galaxy S21", "12.0")
print(android_phone.show_brand()) 
print(android_phone.show_model()) 
print(android_phone.show_version()) 

```

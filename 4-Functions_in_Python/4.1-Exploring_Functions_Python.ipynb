## Introduction to Functions
A function is a reusable block of code designed to perform a specific task.

Functions help organize code, make it modular, and enhance readability and maintainability by allowing tasks to be broken down into smaller, manageable parts.

### Benefits of Using Functions:
__Code Reusability:__ Write once, use multiple times.

__Modularity:__ Break complex tasks into simpler, smaller tasks.

__Organization:__ Structure your code for better readability and debugging.

__Maintainability:__ Simplify updates and modifications by isolating functionality in one place.


### Defining Functions
In Python, functions are defined using the def keyword, followed by the function name, parentheses (), and a colon :. 

```python
# Define Funcation

def func_name():
    print("Hello, I am first function of this Module....")

print("End of Function...")

func_name()
```

### Docstrings
A docstring is a documentation string that describes what the function does. It is optional but highly recommended, especially for complex functions.

```python
def add():
    """ Printing the Addition of Values """
    print( 10 + 20)

add()
```

### Calling Functions

To execute (or "call") a function, simply use its name followed by parentheses () containing any required arguments.

### Function Parameters
Functions often require input values, called parameters, which allow them to process data. You specify parameters in the function definition’s parentheses.

```python
def add(value1, value2):
    """ Printing the Addition of Values """
    print( value1 + value2)

add(10, 30)
add(1, 3)
add(11, 3)
add(66, 11)
add(22, 55)
```

### Default Parameters
You can assign default values to parameters, making them optional when calling the function. If an argument for a default parameter isn’t provided, the default value will be used.

```python
def add(value1=1, value2=2):
    """ Printing the Addition of Values """
    print( value1 + value2)

add(10, 30)
add()
add(50)
```

### Order of Parameters:
When combining default and non-default parameters, non-default parameters should be listed first:

```python
def order_item(item, quantity=1):
    print(f"Ordering {quantity} of {item}.")
```

### Variable-Length Arguments
Python allows you to pass a variable number of arguments to a function. This is useful when you don't know in advance how many arguments a function might need to process.

Types of Variable-Length Arguments:

__Arbitrary Positional Arguments (*args):__ Allows the function to accept a variable number of positional arguments.

__Arbitrary Keyword Arguments (**kwargs):__ Allows the function to accept a variable number of keyword arguments.


```python
# Arbitrary Positional Arguments (*args)

def addition(*args):
    """Return the sum of any number of arguments."""
    print (sum(args))

addition(1,1)
addition(1,10, 45, 51, 43, 76)
addition(1,10, 45, 51, 43, 76, 45, 51, 43, 76)
```

```python
# Arbitrary Keyword Arguments (**kwargs)

def print_user_inform(**kwargs):
    """Print user information passed as keyword arguments."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_user_inform(name="Alice", age=25, country="USA")
```

```python
# Combining *args and **kwargs:

def display_info(required, *args, **kwargs):
    print(f"Required argument: {required}")
    print(f"Additional arguments: {args}")
    print(f"Keyword arguments: {kwargs}")

display_info("Optional1", "Optional2", "Optional3", key1="value1", key2="value2")
```

### Return Statement
The return statement is used to send a result back to the caller. Once return is executed, the function stops and returns the specified value.

```python
def addition(*args):
    """Return the sum of any number of arguments."""
    return sum(args)

sum1 = addition(1,1)
print(sum1)

sum2 = addition(1,10, 45, 51, 43, 76)
print(sum2)
#addition(1,10, 45, 51, 43, 76, 45, 51, 43, 76)
```

### Practical Example: Temperature Conversion


```python
def convert_temperature(temp, unit):
    """This function converts temperature between Celsius and Fahrenheit"""

    ## Celsius To Fahrenheit
    if unit == "C":
        return temp * 9/5 + 32

    ## Fahrenheit to celsius
    elif unit == "F":
        return (temp-32)*5/9
    else:
        return None
    
print(convert_temperature(97.6, 'F'))
print(convert_temperature(100, 'C'))
```

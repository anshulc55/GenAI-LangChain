## <span style="color:yellow"> Exception Handling in Python </span>
Exception handling is a fundamental concept in Python that allows developers to manage errors and unexpected events in code gracefully. 

Instead of crashing, a program can "handle" these errors, making it more robust and user-friendly.

### <span style="color:lightgreen"> Why Use Exception Handling? </span>
1. Ensures program continuity in case of unexpected errors.
2. Provides meaningful error messages to users.
3. Helps in debugging and maintaining code.

### <span style="color:lightgreen"> What is an Exception? </span>
An exception is an event that disrupts the normal flow of a program. In Python, common exceptions include:
1. <span style="color:green"> __ZeroDivisionError__ </span> : Raised when a division by zero occurs.

2. <span style="color:green">__ValueError__</span>: Raised when a function receives an argument of the correct type but inappropriate value.

3. <span style="color:green">__TypeError__</span>: Raised when an operation or function is applied to an object of inappropriate type.

4. <span style="color:green">__IndexError__</span>: Raised when attempting to access an invalid index in a list.

### <span style="color:lightgreen"> Basic Syntax of Exception Handling </span>
In Python, exception handling is achieved using the try, except, else, and finally blocks.

```python
try:
    # Code that may raise an exception
except ExceptionType:
    # Code to handle the exception
else:
    # Code that runs if no exceptions are raised
finally:
    # Code that will always run, regardless of exceptions
```

<span style="color:yellow"> __try block__ </span> : This is where you place the code that might raise an exception.

<span style="color:yellow">__except block__</span>: Code in this block executes if an exception is raised in the try block.

<span style="color:yellow">__else block__</span>: Executes if no exceptions were raised.

<span style="color:yellow">__finally block__</span>: Executes no matter what, useful for cleanup tasks.
